## 2. Load your dataset

```python
df = pd.read_csv("Student_17_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print("Dataset shape (rows, columns):", df.shape)
print("Column names:", list(df.columns))

df.info()

print("\n Missing values in each column:")
display(df.isnull().sum())

print("\n Total Number of duplicate rows:", df.duplicated().sum())

```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
passengers_col = df["Daily_Passengers"]
distance_col = df["Distance_KM"]
# TODO: Convert them to a NumPy array using .to_numpy
passengers_arr = passengers_col.to_numpy()
distance_arr = distance_col.to_numpy()
# TODO: Calculate average, maximum and minimum
print("Average Daily Passengers:", np.mean(passengers_arr))
print("Minimum Daily Passengers:", np.min(passengers_arr))
print("Maximum Daily Passengers:", np.max(passengers_arr))

print("\nAverage Distance (KM):", np.mean(distance_arr))
print("Maximum Distance (KM):", np.max(distance_arr))
print("Minimum Distance (KM):", np.min(distance_arr))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
df["Passengers_Per_KM"] = (df["Daily_Passengers"] / df["Distance_KM"]).round(2)
df["Weekly_Passengers"] = df["Daily_Passengers"]*5
print("Dataset with New Derived Columns")
print(df.head())
# TODO: Filter records using a condition
high_passengers_trips = df[df["Daily_Passengers"] > 60]
print("\n Trips with More Than 60 Daily Passengers")
print(high_passengers_trips)
    
# TODO: Sort records and display the top/bottom results
sorted_by_passengers = df.sort_values(by = "Daily_Passengers", ascending = False)
print("\n Top 5 Highest Passengers Trips")
print(sorted_by_passengers.head())

print("\n Bottom 5 Lowest Passengers Trips")
print(sorted_by_passengers.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate passenger totals by route; find busiest stop; calculate passengers per kilometre.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# TODO: Solve your assigned challenge

# 1. Calculate passenger totals by route
route_totals = df.groupby('Route')['Daily_Passengers'].sum().sort_values(ascending=False)
print(" Total Daily Passengers by Route ")
print(route_totals)

# 2. Find busiest stop
stop_totals = df.groupby('Stop')['Daily_Passengers'].sum().sort_values(ascending=False)
print("\n Total Daily Passengers by Stop ")
print(stop_totals)
print("\nBusiest Stop Overall:", stop_totals.idxmax(), "with", stop_totals.max(), "passengers")

# 3. Calculate passengers per kilometre
df['Passengers_Per_KM'] = (df['Daily_Passengers'] / df['Distance_KM']).round(2)
efficiency_sorted = df.sort_values(by='Passengers_Per_KM', ascending=False)
print("\n Top 5 Most Efficient Trips (Passengers per KM) ")
print(efficiency_sorted[['Route', 'Stop', 'Daily_Passengers', 'Distance_KM', 'Passengers_Per_KM']].head())
```
